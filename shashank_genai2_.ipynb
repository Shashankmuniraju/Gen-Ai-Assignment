
# Assignment 2: Exploring Text Generation Using GPT-2

## Objective
The objective of this assignment is to understand how different text-generation parameters influence the output produced by a language model.

This notebook uses the pretrained **GPT-2** model from the Hugging Face Transformers library. The same opening sentence is used in five experiments while changing one generation parameter at a time.

The experiments cover:
- Temperature
- Top-k
- Top-p
- Repetition penalty

A custom prompt is also used to generate three different versions.


## 1. Install and Import Required Libraries

The notebook uses the Hugging Face Transformers library with the pretrained GPT-2 model. The generation code is organized around a reusable helper function so that each experiment changes only the requested generation setting.

```python
!pip -q install transformers torch
```

```python
import torch
from transformers import pipeline, set_seed

MODEL_NAME = "gpt2"
text_generator = pipeline("text-generation", model=MODEL_NAME)

def generate_continuation(prompt, seed, **generation_options):
    set_seed(seed)
    settings = {
        "max_new_tokens": 80,
        "do_sample": True,
        "pad_token_id": text_generator.tokenizer.eos_token_id,
    }
    settings.update(generation_options)
    result = text_generator(prompt, **settings)
    return result[0]["generated_text"]

print(f"{MODEL_NAME.upper()} model loaded successfully.")
```


## 2. Original Prompt

The same opening sentence is used for the five experiments so that the effect of the changed generation parameter can be compared.


```python
opening_prompt = (
    "The old bookstore at the end of the street had a secret that nobody knew about."
)
print("Original prompt:")
print(opening_prompt)
```

## 3. Experiment 1 - Temperature

**Parameter changed:** Temperature  
**Value:** 0.7

```python
# Experiment setting: temperature = 0.7
generated_text = generate_continuation(opening_prompt, seed=42, temperature=0.7)
print(generated_text)
```

**Observation:** A moderate temperature allows some variation while generally keeping the output coherent.

## 4. Experiment 2 - Top-k

**Parameter changed:** Top-k  
**Value:** 20

```python
# Experiment setting: top-k = 20
generated_text = generate_continuation(opening_prompt, seed=42, top_k=20)
print(generated_text)
```

**Observation:** Restricting choices to the 20 most likely tokens can make the text more focused and predictable.

## 5. Experiment 3 - Top-p

**Parameter changed:** Top-p  
**Value:** 0.85

```python
# Experiment setting: top-p = 0.85
generated_text = generate_continuation(opening_prompt, seed=42, top_p=0.85)
print(generated_text)
```

**Observation:** Top-p provides a probability-based set of candidate tokens and can balance creativity with coherence.

## 6. Experiment 4 - Repetition Penalty

**Parameter changed:** Repetition penalty  
**Value:** 1.5

```python
# Experiment setting: repetition penalty = 1.5
generated_text = generate_continuation(opening_prompt, seed=42, repetition_penalty=1.5)
print(generated_text)
```

**Observation:** A higher repetition penalty discourages the model from repeatedly selecting the same tokens or phrases.

## 7. Experiment 5 - Lower Temperature

**Parameter changed:** Temperature  
**Value:** 0.3

```python
# Experiment setting: temperature = 0.3
generated_text = generate_continuation(opening_prompt, seed=42, temperature=0.3)
print(generated_text)
```

**Observation:** A lower temperature makes the generation more conservative and predictable.

## 8. Comparison of the Five Experiments

| Experiment | Parameter Changed | Value | Main Effect |
|---|---|---:|---|
| 1 | Temperature | 0.7 | Moderate creativity and coherence |
| 2 | Top-k | 20 | More focused token choices |
| 3 | Top-p | 0.85 | Balance between variation and coherence |
| 4 | Repetition penalty | 1.5 | Reduced repetition |
| 5 | Temperature | 0.3 | More predictable and conservative text |

### Overall Analysis
GPT-2 uses these parameters to control how it selects the next token. A lower temperature generally makes the output more predictable, whereas a higher temperature permits more variation. Top-k keeps only a fixed number of high-probability choices, while top-p keeps a probability-based candidate set. Repetition penalty reduces the tendency to reuse tokens.

For predictable writing, a lower temperature is useful. Moderate or higher temperature can increase creativity, although too much randomness may make the text less coherent.


## 9. Custom Prompt

A short product-description prompt is used for the custom generation task. Three versions are generated with different temperature values.


```python
custom_prompt = (
    "Introducing the SmartBottle, a reusable bottle designed for busy students "
    "who want to stay hydrated throughout the day."
)
print("Custom prompt:")
print(custom_prompt)
```

### Custom Generation 1

```python
custom_text = generate_continuation(
    custom_prompt,
    seed=10,
    temperature=0.5,
    top_p=0.9,
    max_new_tokens=60,
)
print(custom_text)
```

**Temperature:** 0.5

### Custom Generation 2

```python
custom_text = generate_continuation(
    custom_prompt,
    seed=20,
    temperature=0.7,
    top_p=0.9,
    max_new_tokens=60,
)
print(custom_text)
```

**Temperature:** 0.7

### Custom Generation 3

```python
custom_text = generate_continuation(
    custom_prompt,
    seed=30,
    temperature=0.9,
    top_p=0.9,
    max_new_tokens=60,
)
print(custom_text)
```

**Temperature:** 0.9

## 10. Custom Generation Observations

The custom prompt is kept constant while the temperature changes across the three versions.

- **Version 1 (0.5):** More controlled and predictable.
- **Version 2 (0.7):** A balance between creativity and coherence.
- **Version 3 (0.9):** More varied and creative, with a higher chance of unusual wording.

Because sampling is enabled, exact text can vary between environments. The fixed seeds make the examples more reproducible.

## Conclusion

The experiments show that GPT-2 can produce noticeably different text when its generation settings are adjusted. Lower or more restrictive settings generally make the output more focused and predictable, while increasing randomness can make the text more varied but may also reduce coherence. Top-k and top-p control how many likely next tokens are considered, whereas repetition penalty helps reduce repeated patterns. The custom-prompt experiments also demonstrate that the same language model can create different responses from the same starting idea. Overall, the activity shows that generation parameters have an important effect on the balance between creativity, predictability, and text quality.
