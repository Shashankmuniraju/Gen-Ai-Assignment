```python
import torch
import pickle
import numpy as np
from PIL import Image
import os

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

with open("ffhq.pkl", "rb") as f:
    G = pickle.load(f)["G_ema"].to(device)

G.eval()

print("Latent dimension:", G.z_dim)

# Two randomly selected latent vectors
torch.manual_seed(42)

z1 = torch.randn([1, G.z_dim], device=device)
z2 = torch.randn([1, G.z_dim], device=device)

# Generate images
with torch.no_grad():
    img1 = G(z1, None, truncation_psi=0.7)
    img2 = G(z2, None, truncation_psi=0.7)

def tensor_to_image(img):
    img = (img.clamp(-1, 1) + 1) * 127.5
    img = img.permute(0, 2, 3, 1)
    img = img[0].cpu().numpy().astype(np.uint8)
    return Image.fromarray(img)

image1 = tensor_to_image(img1)
image2 = tensor_to_image(img2)

os.makedirs("assignment5", exist_ok=True)

image1.save("assignment5/image_z1.png")
image2.save("assignment5/image_z2.png")

display(image1)
display(image2)
```

```python
import torch
import numpy as np
from PIL import Image
import os

num_images = 12

interpolated_images = []

for i in range(num_images):

    t = i / (num_images - 1)

    # Linear interpolation
    z = (1 - t) * z1 + t * z2

    with torch.no_grad():
        generated = G(z, None, truncation_psi=0.7)

    img = tensor_to_image(generated)

    interpolated_images.append(img)

    img.save(f"assignment5/interpolation_{i+1:02d}.png")

    print(f"Generated image {i+1}/{num_images}")
```

```python
gif_images = []

for img in interpolated_images:
    gif_images.append(img.resize((256, 256)))

gif_images[0].save(
    "assignment5/stylegan2_interpolation.gif",
    save_all=True,
    append_images=gif_images[1:],
    duration=500,
    loop=0
)

print("GIF created successfully!")
```

```python
z_original = z1.clone()

dimension = 0

values = [-2, -1, 0, 1, 2]

dimension_images = []

for value in values:

    z_modified = z_original.clone()

    # Change ONLY one latent dimension
    z_modified[0, dimension] = value

    with torch.no_grad():
        generated = G(
            z_modified,
            None,
            truncation_psi=0.7
        )

    img = tensor_to_image(generated)
    dimension_images.append(img)

    img.save(
        f"assignment5/dimension_{dimension}_{value}.png"
    )

display(dimension_images[2])
```
